This is the line we use to import the pandas library into our program:

```python
import pandas as pd
```

pandas stores a table in an object called a **DataFrame**. Like a spreadsheet table, a DataFrame has rows and columns, and each column has a name. The difference is that you work with it through code rather than by clicking on cells.

The code cell below reads the CSV file and stores the resulting DataFrame under the name `df`, a common short name for "DataFrame". (You could use any name you like.)

For `read_csv` to work, the notebook needs to be able to find the file. That means you need to load the file into your Colab session. The course gives you instructions for doing this.

```python
df = pd.read_csv("sydney_pm25_wide.csv")
```

You won't scroll through a DataFrame cell by cell as you would in a spreadsheet. Instead, you ask it questions, such as how many rows it has, what its columns are called, or what the first few rows look like.

That's what the code cell below does: the `head()` method essentially extracts the first five rows (after the header row) from the DataFrame.

In Excel or Google Sheets, that would look something like this:

first five rows in Google Sheets.png

Run the code cell below, and notice how `df.head()` essentially gives you the same information:


```python
print(df.head())
```

`df.tail()` is similar to `df.head()`, except it shows you the **last** five rows in the table:

```python
print(df.tail())
```

`df.shape` is an attribute of the df DataFrame. It tells you the number of rows and columns in the table (not counting the header row). For example, an output like this:

`(500,4)`

means that there are 500 rows of data, and 4 columns.


```python
print(df.shape)
```

And finally, `df.columns` is another attribute of the DataFrame. It tells you the names of the columns in the table.

Note: This also captures the data type (`dtype='object'`) of the whole DataFrame. We don't need to pay attention to that for now. When you run the code cell below, see if you can spot the column names and match them up to what you see in Excel or Google Sheets:

```python
print(df.columns)
```
