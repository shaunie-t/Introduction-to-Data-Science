The following code cell contains one week (7 days) of daily **PM2.5 readings**.

PM2.5 readings measure the concentration of fine inhalable air particles that have a diameter of 2.5 micrometres or less. These readings are usually reported in micrograms per cubic metre (µg/m³). For example, a reading of 6.2 means that there are 6.2 micrograms of these fine air particles per cubic metre.

A high PM2.5 reading means more of these fine particles, which is often bad for human health as they can stay in the lungs and get into the bloodstream.

```python
readings = [6.2, 8.1, 7.5, 25.4, 31.0, 9.8, 5.9]

threshold = 25
```

The following code cell contains some calculations based on the daily values and the threshold above.

```python
average = sum(readings) / len(readings)

days_above = len([r for r in readings if r > threshold])

print(f"Average PM2.5: {average:.1f}")

print(f"Days above {threshold}: {days_above}")
```
